```python
fortune = pd.read_csv('/Users/MarcelMiłkowski/OneDrive - DIGITAL TPA/moje/course-files/course-files/Fortune_500_2017.csv', usecols=["Rank","Title","Industry","Hqlocation","Sector"])

fortune["Sector"] = fortune["Sector"].str.upper()
fortune

fortune["Industry"].str.lower().str.contains("comp")

fortune = fortune.rename(columns = {"Hqlocation" : "HQLOCATION"})  # ZMIANA NA PRZYPOMNIENIE

fortune.loc[35,"HQLOCATION"] = "cinicii, OHIO, BORUTA"

expanded = fortune["HQLOCATION"].str.strip().str.split(", ", expand=True, n=0)
fortune[["City", "State", "test"]] = expanded

fortune.drop(columns="HQLOCATION", inplace=True)
#fortune["test"] = fortune["test"].fillna("?")
#fortune[fortune["test"] != "?"]

frame = fortune.copy()

def BuildShortcut(row):
    industry = row["Industry"]
    result = ""
    for i in industry.split(" "):
        result += i[0]
    return result

fortune["shortcut"] = fortune.apply(BuildShortcut, axis = 1)
fortune

def BuildTitleLenght(row):
    title = row["Title"]
    space = -1
    for i in title.split():
        space += 1

    text = f'ilość słow, {len(title.split())}, ilosc spacji: {space}'
    return text

fortune["Title Info"] = fortune.apply(BuildTitleLenght, axis=1)
fortune

def laczenie(column):
    column = column.fillna("").astype(str)
    jeden_dlugi = " ".join(column)
    return jeden_dlugi

wiersz =  fortune.apply(laczenie, axis= 0)
fortune = pd.concat([fortune, pd.DataFrame([wiersz])], ignore_index=True)
fortune.tail(3)

pd.DataFrame([wiersz])

type([wiersz])






```
