```python
def NameLenght(row):
    return len(row["Name"])

pokemon["NameLength"] = pokemon.apply(NameLenght, axis=1)
pokemon.sort_values(by="NameLength")

def HasSecondType(row):
    type2 = row["Type 2"]
    if pd.isna(type2):
        return False
    else:
        return True
    
pokemon["HasSecondType"] = pokemon.apply(HasSecondType, axis = 1)
pokemon

pokemon["DefenseTotal"] = pokemon["Defense"] + pokemon["Sp. Def"]
pokemon

def IsFast(row):       ## CIEKAWE
    return row["Speed"] > 100

pokemon["IsFast"] = pokemon.apply(IsFast, axis=1)
pokemon[pokemon.apply(IsFast, axis=1)]



def Classes(value):
    if value > 650:
        return "Strong"
    if value > 350:
        return "Average"
    else:
        return "Weak"

def PowerLabel(row):
    pok = Classes(row["Total"])
    label = f'{pok}'
    return label

pokemon["Label"] = pokemon.apply(PowerLabel, axis=1)

def Legendary(row):
    zz = row["Legendary"]
    if zz:
        return "Legendary"
    else:
        return "Normal"
    
LegendaryColumn = pokemon.apply(Legendary, axis = 1)
pokemon.insert(12, "LegendaryNormalized", LegendaryColumn)
isWeak = (pokemon["Label"] == "Strong")
StrongPokemon = pokemon[isWeak].copy()
StrongPokemon
pokemon



```
